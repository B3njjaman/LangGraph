## Esquema

Cuando definimos un `StateGraph` de LangGraph, usamos un [esquema del estado](https://docs.langchain.com/oss/python/langgraph/graph-api/#state).

El esquema del estado representa la estructura y los tipos de los datos que usará nuestro grafo.

Se espera que todos los nodos se comuniquen con ese esquema.

LangGraph es flexible para definir el esquema del estado: acepta distintos [tipos](https://docs.python.org/3/library/stdtypes.html#type-objects) de Python y distintas formas de validación.

## TypedDict

Como vimos en el módulo 1, podemos usar la clase `TypedDict` del módulo `typing` de Python.

Permite especificar las claves y el tipo de valor de cada una.

Pero ten en cuenta que son anotaciones de tipo.

Los verificadores estáticos de tipos (como [mypy](https://github.com/python/mypy)) o los IDE pueden usarlas para detectar errores de tipos antes de ejecutar el código.

Pero no se hacen cumplir en tiempo de ejecución.

```python
from typing_extensions import TypedDict

class EstadoTypedDict(TypedDict):
    foo: str
    bar: str
```

Para restringir más los valores, puedes usar anotaciones como `Literal`.

Aquí, `animo` solo puede ser "feliz" o "triste".

```python
from typing import Literal

class EstadoTypedDict(TypedDict):
    nombre: str
    animo: Literal["feliz","triste"]
```

Para usar la clase de estado que definimos (aquí, `EstadoTypedDict`) en LangGraph, basta con pasársela a `StateGraph`.

Y podemos pensar en cada clave del estado como un "canal" de nuestro grafo.

Como vimos en el módulo 1, en cada nodo sobrescribimos el valor de una clave o "canal" determinado.

```python
import random
from IPython.display import Image, display
from langgraph.graph import StateGraph, START, END

def nodo_1(estado):
    print("---Nodo 1---")
    return {"nombre": estado['nombre'] + " está ... "}

def nodo_2(estado):
    print("---Nodo 2---")
    return {"animo": "feliz"}

def nodo_3(estado):
    print("---Nodo 3---")
    return {"animo": "triste"}

def decidir_animo(estado) -> Literal["nodo_2", "nodo_3"]:

    # Aquí hacemos un reparto 50 / 50 entre los nodos 2 y 3
    if random.random() < 0.5:

        # El 50% de las veces devolvemos el nodo 2
        return "nodo_2"

    # El 50% de las veces devolvemos el nodo 3
    return "nodo_3"

# Construye el grafo
constructor = StateGraph(EstadoTypedDict)
constructor.add_node("nodo_1", nodo_1)
constructor.add_node("nodo_2", nodo_2)
constructor.add_node("nodo_3", nodo_3)

# Lógica
constructor.add_edge(START, "nodo_1")
constructor.add_conditional_edges("nodo_1", decidir_animo)
constructor.add_edge("nodo_2", END)
constructor.add_edge("nodo_3", END)

# Compila
grafo = constructor.compile()

# Visualiza
display(Image(grafo.get_graph().draw_mermaid_png()))
```

Como nuestro estado es un diccionario, basta con invocar el grafo con un diccionario para fijar el valor inicial de la clave `nombre` del estado.

```python
grafo.invoke({"nombre":"Lance"})
```

## Dataclass

Las [dataclasses](https://docs.python.org/3/library/dataclasses.html) de Python ofrecen [otra forma de definir datos estructurados](https://www.datacamp.com/tutorial/python-data-classes).

Las dataclasses tienen una sintaxis concisa para crear clases pensadas sobre todo para guardar datos.

```python
from dataclasses import dataclass

@dataclass
class EstadoDataclass:
    nombre: str
    animo: Literal["feliz","triste"]
```

Para acceder a las claves de una `dataclass`, solo hay que cambiar la forma de acceder en `nodo_1`:

* Usamos `estado.nombre` para el estado `dataclass`, en lugar de `estado["nombre"]` como en el `TypedDict` de antes

Notarás algo un poco raro: en cada nodo seguimos devolviendo un diccionario para actualizar el estado.

Esto es posible porque LangGraph guarda cada clave de tu objeto de estado por separado.

El objeto que devuelve el nodo solo necesita tener claves (atributos) que coincidan con las del estado.

En este caso, la `dataclass` tiene la clave `nombre`, así que podemos actualizarla pasando un diccionario desde el nodo, igual que cuando el estado era un `TypedDict`.

```python
def nodo_1(estado):
    print("---Nodo 1---")
    return {"nombre": estado.nombre + " está ... "}

# Construye el grafo
constructor = StateGraph(EstadoDataclass)
constructor.add_node("nodo_1", nodo_1)
constructor.add_node("nodo_2", nodo_2)
constructor.add_node("nodo_3", nodo_3)

# Lógica
constructor.add_edge(START, "nodo_1")
constructor.add_conditional_edges("nodo_1", decidir_animo)
constructor.add_edge("nodo_2", END)
constructor.add_edge("nodo_3", END)

# Compila
grafo = constructor.compile()

# Visualiza
display(Image(grafo.get_graph().draw_mermaid_png()))
```

Invocamos con una `dataclass` para fijar los valores iniciales de cada clave / canal de nuestro estado.

```python
grafo.invoke(EstadoDataclass(nombre="Lance",animo="triste"))
```

## Pydantic

Como dijimos, `TypedDict` y las `dataclasses` ofrecen anotaciones de tipo, pero no hacen cumplir los tipos en tiempo de ejecución.

Esto significa que podrías asignar valores no válidos sin que salte ningún error.

Por ejemplo, podemos fijar `animo` en `enojado` aunque nuestra anotación diga `animo: list[Literal["feliz","triste"]]`.

```python
instancia_dataclass = EstadoDataclass(nombre="Lance", animo="enojado")
```

[Pydantic](https://docs.pydantic.dev/latest/api/base_model/) es una biblioteca de validación de datos y de manejo de configuración que usa anotaciones de tipo de Python.

Es especialmente adecuada [para definir esquemas del estado en LangGraph](https://docs.langchain.com/oss/python/langgraph/use-graph-api#use-pydantic-models-for-graph-state) gracias a su capacidad de validación.

Pydantic puede validar, en tiempo de ejecución, que los datos cumplan los tipos y restricciones especificados.

```python
from pydantic import BaseModel, field_validator, ValidationError

class EstadoPydantic(BaseModel):
    nombre: str
    animo: str # "feliz" o "triste"

    @field_validator('animo')
    @classmethod
    def validar_animo(cls, valor):
        # Asegura que el ánimo sea "feliz" o "triste"
        if valor not in ["feliz", "triste"]:
            raise ValueError("Cada ánimo debe ser 'feliz' o 'triste'")
        return valor

try:
    estado = EstadoPydantic(nombre="John Doe", animo="enojado")
except ValidationError as e:
    print("Error de validación:", e)
```

Podemos usar `EstadoPydantic` en nuestro grafo sin problemas. 

```python
# Construye el grafo
constructor = StateGraph(EstadoPydantic)
constructor.add_node("nodo_1", nodo_1)
constructor.add_node("nodo_2", nodo_2)
constructor.add_node("nodo_3", nodo_3)

# Lógica
constructor.add_edge(START, "nodo_1")
constructor.add_conditional_edges("nodo_1", decidir_animo)
constructor.add_edge("nodo_2", END)
constructor.add_edge("nodo_3", END)

# Compila
grafo = constructor.compile()

# Visualiza
display(Image(grafo.get_graph().draw_mermaid_png()))
```

```python
grafo.invoke(EstadoPydantic(nombre="Lance",animo="triste"))
```
